```python
import pandas as pd
import numpy as np
```

```python
data=pd.read_csv('car-sales-extended.csv')
print('car price estimator')
data.isna().sum()
```

```python
x= data.drop('Price',axis=1)
y= data['Price']
```

```python
#split training and testing datasets
np.random.seed(42)
import sklearn.model_selection as splitter
xtrain,xtest,ytrain,ytest=splitter.train_test_split(x,y,test_size=0.2)
```

```python
import sklearn.preprocessing as encode
import sklearn.compose as transform
encoder=encode.OneHotEncoder()

#set category names
categorical_features=["Make","Colour","Doors"]

#transform data
transformed=transform.ColumnTransformer([("encoded",encoder,categorical_features)],remainder='passthrough')

#fit the data
transformeddata=transformed.fit_transform(xtrain)
transformeddata
transformeddata2=transformed.fit_transform(xtest)
```

```python
import sklearn.ensemble as cl
learner=cl.RandomForestRegressor(n_estimators=20)
learner.fit(transformeddata,ytrain)
pred=learner.predict(transformeddata)
learner.score(transformeddata,ytrain)*100
```
